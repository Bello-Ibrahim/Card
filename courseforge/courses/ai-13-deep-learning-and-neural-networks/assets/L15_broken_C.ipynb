## Setup
Imports, a fixed random seed and the data.

```python
import torch
from torch import nn

torch.manual_seed(0)
print("PyTorch", torch.__version__)
from torch.utils.data import TensorDataset, DataLoader

# Synthetic 10-class data: 2,048 examples with 64 features, labels from a fixed random linear rule
X = torch.randn(2048, 64)
W_true = torch.randn(64, 10)
y = (X @ W_true).argmax(1)
train_dl = DataLoader(TensorDataset(X, y), batch_size=64, shuffle=True)
loss_fn = nn.CrossEntropyLoss()
```

## Training code (broken)
Run this cell unchanged first and note the symptom.

```python
# --- Code as given in the lesson ---
layers = []
for _ in range(20): layers += [nn.Linear(64, 64), nn.Sigmoid()]
model = nn.Sequential(*layers, nn.Linear(64, 10))

# --- Training loop ---
opt = torch.optim.SGD(model.parameters(), lr=0.05)
for epoch in range(8):
    total, n = 0.0, 0
    for xb, yb in train_dl:
        opt.zero_grad()
        loss = loss_fn(model(xb), yb)
        loss.backward()
        opt.step()
        total += loss.item() * len(xb); n += len(xb)
    print(f"epoch {epoch + 1}  mean loss {total / n:.4f}")
```
