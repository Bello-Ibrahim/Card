## Training code (broken)
Run this cell unchanged first and note the symptom.

```python
# --- Code as given in the lesson ---
X = torch.rand(512, 5) * 5000                   # raw values in the thousands
y = X @ torch.tensor([0.2, 0.1, 0.05, 0.3, 0.1])
model = nn.Linear(5, 1); opt = torch.optim.SGD(model.parameters(), lr=0.01)

# --- Training loop ---
for step in range(20):
    opt.zero_grad()
    loss = loss_fn(model(X).squeeze(1), y)
    loss.backward()
    opt.step()
    print(f"step {step:2d}  loss {loss.item():.4g}")
```
